**Título:** Balanço Geral de Massa e Energia em Reatores Químicos Contínuos (CSTR)

> **Enunciado Original:**
> Determinar a vazão de fluido de resfriamento necessária na camisa de um reator químico contínuo CSTR operando em estado estacionário com reação exotérmica de Delta H = -85 kJ/mol.

### Formulação Matemática Formal
$$\frac{dE_{vc}}{dt} = \dot{Q} - \dot{W} + \sum \dot{m}_{in} h_{in} - \sum \dot{m}_{out} h_{out} + r_A V (-\Delta H_R) = 0$$


### Passo 1: Geração Térmica da Reação Exotérmica
\dot{Q}_{gerado} = r_A V (-\Delta H_R) = (0,05 \text{ mol/L}\cdot\text{s}) \times (1000 \text{ L}) \times (85.000 \text{ J/mol}) = 4.250.000 \text{ W} = 4,25 \text{ MW}.


### Passo 2: Calor Removido pelo Efluente de Saída
\dot{Q}_{escoamento} = \dot{m} C_p (T_{out} - T_{in}) = 10 \text{ kg/s} \times 4180 \times (80 - 25) = 2.299.000 \text{ W} = 2,30 \text{ MW}.


### Passo 3: Carga Térmica a ser Absorvida pela Camisa
\dot{Q}_{camisa} = \dot{Q}_{gerado} - \dot{Q}_{escoamento} = 4,25 - 2,30 = 1,95 \text{ MW}.


### Passo 4: Vazão Mássica de Água na Camisa de Resfriamento
\dot{m}_{água} = \frac{\dot{Q}_{camisa}}{C_p \Delta T_{camisa}} = \frac{1.950.000}{4180 \times (35 - 15)} = \frac{1.950.000}{83.600} \approx 23,32 \text{ kg/s}.


### Passo 5: Conclusão Operacional
Balanço estacionário rigorosamente fechado com controle PID de vazão evitando fuga térmica (runaway reaction).


```python
import numpy as np
import matplotlib.pyplot as plt

vazao_agua = np.linspace(10, 40, 50)
delta_t = 1950000 / (vazao_agua * 4180)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f06')
ax.set_facecolor('#210f06')
ax.plot(vazao_agua, delta_t, color='#ea580c', lw=2.5, label='Elevação Térmica na Camisa ΔT (°C)')
ax.axhline(20, color='#10b981', linestyle='--', label='Meta Operacional (ΔT = 20 °C -> 23.3 kg/s)')
ax.set_title('Elevação Térmica na Camisa vs Vazão de Resfriamento', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Vazão Mássica de Água (kg/s)', color='#94a3b8')
ax.set_ylabel('Elevação de Temperatura ΔT (°C)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2510', alpha=0.5)
ax.legend(facecolor='#33180a', edgecolor='#ea580c', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
