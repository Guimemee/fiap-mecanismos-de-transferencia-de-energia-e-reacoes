> **Enunciado Original:**
> Calcular o número de Reynolds, o número de Nusselt e o coeficiente de convecção h para escoamento de água a 60 °C no interior de um tubo com diâmetro D = 25 mm a uma velocidade de v = 1.5 m/s.

### Formulação Matemática Formal
$$Re_D = \frac{\rho v D}{\mu}, \quad Nu_D = 0,023 Re_D^{0,8} Pr^{0,4} \quad (\text{Dittus-Boelter})$$


### Passo 1: Propriedades Físicas da Água a 60 °C
\rho = 983 \text{ kg/m}^3, \quad \mu = 4,67 \times 10^{-4} \text{ Pa}\cdot\text{s}, \quad k = 0,654 \text{ W/mK}, \quad Pr = 2,98.


### Passo 2: Cálculo do Número de Reynolds
Re_D = \frac{983 \times 1,5 \times 0,025}{4,67 \times 10^{-4}} = \frac{36,8625}{4,67 \times 10^{-4}} \approx 78.935 \text{ (Regime Turbulento Pleno)}.


### Passo 3: Correlação de Dittus-Boelter para Aquecimento
Nu_D = 0,023 \times (78.935)^{0,8} \times (2,98)^{0,4} = 0,023 \times 8.243 \times 1,546 \approx 293,1.


### Passo 4: Coeficiente Médio de Transferência Convectiva
h = Nu_D \frac{k}{D} = 293,1 \times \frac{0,654}{0,025} \approx 7.667 \text{ W/m}^2\cdot\text{K}.


### Passo 5: Conclusão Hidráulica
Elevado coeficiente convectivo assegurando resfriamento ultrarrápido dos reatores em regime turbulento.


```python
import numpy as np
import matplotlib.pyplot as plt

v = np.linspace(0.2, 3.0, 50)
re = (983 * v * 0.025) / 4.67e-4
nu = 0.023 * (re**0.8) * (2.98**0.4)
h = nu * (0.654 / 0.025)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f06')
ax.set_facecolor('#210f06')
ax.plot(v, h, color='#f97316', lw=2.5, label='Coeficiente h (W/m²K)')
ax.set_title('Coeficiente de Convecção vs Velocidade de Escoamento', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Velocidade do Fluido (m/s)', color='#94a3b8')
ax.set_ylabel('Coeficiente Convectivo h (W/m²K)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2510', alpha=0.5)
ax.legend(facecolor='#33180a', edgecolor='#ea580c', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
