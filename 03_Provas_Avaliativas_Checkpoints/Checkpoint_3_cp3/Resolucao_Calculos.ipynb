**Título:** Radiação Térmica: Lei de Stefan-Boltzmann e Fatores de Forma

> **Enunciado Original:**
> Calcular a taxa líquida de transferência de calor por radiação entre a carcaça de um reator a T_1 = 500 K (emissividade epsilon = 0.85, área A = 2.0 m^2) e as paredes da sala a T_2 = 300 K.

### Formulação Matemática Formal
$$q_{rad} = \epsilon \sigma A (T_1^4 - T_2^4), \quad \sigma = 5,670 \times 10^{-8} \text{ W/m}^2\cdot\text{K}^4$$


### Passo 1: Aplicação da Lei de Stefan-Boltzmann para Superfícies Cinzentas
q_{rad} = \epsilon \sigma A (T_1^4 - T_2^4).


### Passo 2: Cálculo das Quartas Potências de Temperatura Absoluta
T_1^4 = (500)^4 = 6,25 \times 10^{10} \text{ K}^4 \quad \text{e} \quad T_2^4 = (300)^4 = 8,10 \times 10^9 \text{ K}^4.


### Passo 3: Diferença de Potências Térmicas
T_1^4 - T_2^4 = 62,5 \times 10^9 - 8,1 \times 10^9 = 54,4 \times 10^9 \text{ K}^4.


### Passo 4: Resolução Numérica do Calor Irradiado
q_{rad} = 0,85 \times (5,670 \times 10^{-8}) \times 2,0 \times (54,4 \times 10^9) = 0,09639 \times 54,4 \times 10 = 5.243,6 \text{ Watts}.


### Passo 5: Conclusão Radiativa
A dissipação por radiação corresponde a 5,24 kW, tornando-se comparável à convecção natural em altas temperaturas.


```python
import numpy as np
import matplotlib.pyplot as plt

t1 = np.linspace(320, 600, 100)
q_rad = 0.85 * 5.67e-8 * 2.0 * (t1**4 - 300**4)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f06')
ax.set_facecolor('#210f06')
ax.plot(t1 - 273.15, q_rad / 1e3, color='#ea580c', lw=2.5, label='Potência Irradiada (kW)')
ax.set_title('Calor Transferido por Radiação vs Temperatura da Carcaça', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Temperatura da Superfície (°C)', color='#94a3b8')
ax.set_ylabel('Potência Radiativa q (kW)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2510', alpha=0.5)
ax.legend(facecolor='#33180a', edgecolor='#ea580c', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
