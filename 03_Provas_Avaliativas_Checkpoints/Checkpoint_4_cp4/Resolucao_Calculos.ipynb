**Título:** Dimensionamento de Trocador de Calor pelo Método LMTD (Média Logarítmica)

> **Enunciado Original:**
> Dimensionar a área de troca térmica A de um trocador de calor casco e tubos em contracorrente para resfriar óleo de 120 °C para 50 °C usando água de 20 °C para 40 °C com U = 450 W/m^2K.

### Formulação Matemática Formal
$$\Delta T_{ln} = \frac{\Delta T_1 - \Delta T_2}{\ln(\Delta T_1 / \Delta T_2)}, \quad A = \frac{q}{U \cdot \Delta T_{ln}}$$


### Passo 1: Balanço de Calor no Fluido Quente
q = \dot{m} C_p (T_{h,in} - T_{h,out}) = 2,0 \text{ kg/s} \times 2100 \text{ J/kgK} \times (120 - 50) = 294.000 \text{ W} = 294 \text{ kW}.


### Passo 2: Deltas de Temperatura em Contracorrente
\Delta T_1 = T_{h,in} - T_{c,out} = 120 - 40 = 80 ^\circ\text{C}; \quad \Delta T_2 = T_{h,out} - T_{c,in} = 50 - 20 = 30 ^\circ\text{C}.


### Passo 3: Média Logarítmica das Diferenças de Temperatura (LMTD)
\Delta T_{ln} = \frac{80 - 30}{\ln(80 / 30)} = \frac{50}{\ln(2,667)} = \frac{50}{0,9808} \approx 50,98 ^\circ\text{C}.


### Passo 4: Cálculo da Área de Troca Térmica
A = \frac{294.000}{450 \times 50,98} = \frac{294.000}{22.941} \approx 12,82 \text{ m}^2.


### Passo 5: Conclusão de Projeto
Configuração com feixe de 48 tubos de 1,5 polegadas em casco único de 3 metros de comprimento.


```python
import numpy as np
import matplotlib.pyplot as plt

posicao = np.array([0, 0.5, 1.0])
t_quente = np.array([120, 85, 50])
t_frio = np.array([40, 30, 20])
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f06')
ax.set_facecolor('#210f06')
ax.plot(posicao, t_quente, 'o-', color='#ef4444', lw=2.5, label='Fluido Quente (Óleo)')
ax.plot(posicao, t_frio, 's-', color='#38bdf8', lw=2.5, label='Fluido Frio em Contracorrente (Água)')
ax.set_title('Perfil de Temperatura no Trocador Contracorrente', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Posição Adimensional ao Longo do Tubo (x/L)', color='#94a3b8')
ax.set_ylabel('Temperatura (°C)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2510', alpha=0.5)
ax.legend(facecolor='#33180a', edgecolor='#ea580c', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
