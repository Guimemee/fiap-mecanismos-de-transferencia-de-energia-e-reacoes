**Título:** Condução Térmica Unidimensional em Paredes Compostas e Isolamento

> **Enunciado Original:**
> Calcular a taxa de transferência de calor por condução q e a temperatura na interface de uma parede composta de reator farmacêutico formada por aço inox (k_1 = 16 W/mK, L_1 = 10 mm) e isolante de lã de rocha (k_2 = 0.04 W/mK, L_2 = 50 mm).

### Formulação Matemática Formal
$$R_{total} = \frac{L_1}{k_1 A} + \frac{L_2}{k_2 A}, \quad q = \frac{T_{in} - T_{out}}{R_{total}}$$


### Passo 1: Resistência Térmica por Condução do Aço Inox
R_1 = \frac{L_1}{k_1} = \frac{0,010}{16} = 0,000625 \text{ m}^2\cdot\text{K/W}.


### Passo 2: Resistência Térmica do Isolante Lã de Rocha
R_2 = \frac{L_2}{k_2} = \frac{0,050}{0,04} = 1,250000 \text{ m}^2\cdot\text{K/W}.


### Passo 3: Resistência Térmica Condutiva Total
R_{total} = R_1 + R_2 = 1,250625 \text{ m}^2\cdot\text{K/W} (o isolante representa 99,95% da resistência total).


### Passo 4: Fluxo de Calor Específico para Tin = 180 °C e Tout = 30 °C
q'' = \frac{180 - 30}{1,250625} = \frac{150}{1,250625} \approx 119,94 \text{ W/m}^2.


### Passo 5: Temperatura na Interface Aço-Isolante
T_{interface} = T_{in} - q'' R_1 = 180 - (119,94 \times 0,000625) \approx 179,92 ^\circ\text{C}.


### Passo 6: Conclusão Térmica
O isolante térmico reduz a perda de calor em mais de 98% garantindo estabilidade química da reação farmacêutica.


```python
import numpy as np
import matplotlib.pyplot as plt

espessura_mm = np.linspace(10, 100, 50)
r_isol = (espessura_mm * 1e-3) / 0.04
q_fluxo = 150 / (0.000625 + r_isol)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f06')
ax.set_facecolor('#210f06')
ax.plot(espessura_mm, q_fluxo, color='#ea580c', lw=2.5, label='Fluxo de Calor Residual (W/m²)')
ax.scatter([50], [119.94], color='#10b981', s=60, zorder=5, label='Projeto (50 mm -> 120 W/m²)')
ax.set_title('Perda de Calor vs Espessura de Isolamento', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Espessura do Isolante (mm)', color='#94a3b8')
ax.set_ylabel('Fluxo de Calor q" (W/m²)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2510', alpha=0.5)
ax.legend(facecolor='#33180a', edgecolor='#ea580c', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
