> **Enunciado Original:**
> Determinar a energia de ativação E_a e o fator pré-exponencial A para uma reação de degradação térmica de fármaco medindo as constantes de velocidade k_1 e k_2 nas temperaturas T_1 = 300 K e T_2 = 340 K.

### Formulação Matemática Formal
$$k(T) = A e^{-\frac{E_a}{R T}}, \quad \ln\left(\frac{k_2}{k_1}\right) = \frac{E_a}{R}\left(\frac{1}{T_1} - \frac{1}{T_2}\right)$$


### Passo 1: Constantes de Velocidade Experimentais
k_1(300 \text{ K}) = 0,0024 \text{ s}^{-1} \quad \text{e} \quad k_2(340 \text{ K}) = 0,0860 \text{ s}^{-1}.


### Passo 2: Razão Logarítmica de Velocidades
\ln\left(\frac{0,0860}{0,0024}\right) = \ln(35,83) \approx 3,579.


### Passo 3: Diferença Inversa de Temperatura
\frac{1}{T_1} - \frac{1}{T_2} = \frac{1}{300} - \frac{1}{340} = 0,003333 - 0,002941 = 0,000392 \text{ K}^{-1}.


### Passo 4: Cálculo da Energia de Ativação
E_a = \frac{3,579 \times 8,314}{0,000392} = \frac{29,756}{0,000392} \approx 75.908 \text{ J/mol} = 75,91 \text{ kJ/mol}.


### Passo 5: Conclusão Cinética
A sensibilidade térmica da molécula exige rigoroso controle em câmara frigorífica a 4 °C para prazo de validade de 24 meses.


```python
import numpy as np
import matplotlib.pyplot as plt

t_k = np.linspace(290, 360, 50)
ea = 75908
r = 8.314
k = 3.8e10 * np.exp(-ea / (r * t_k))
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f06')
ax.set_facecolor('#210f06')
ax.plot(t_k - 273.15, k, color='#ea580c', lw=2.5, label='Constante k(T) (s⁻¹)')
ax.set_yscale('log')
ax.set_title('Equação de Arrhenius: Taxa de Reação vs Temperatura', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Temperatura (°C)', color='#94a3b8')
ax.set_ylabel('Constante de Velocidade k (Escala Log)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2510', alpha=0.5)
ax.legend(facecolor='#33180a', edgecolor='#ea580c', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
